```python
import nltk
from nltk.tokenize import word_tokenize, WhitespaceTokenizer, TreebankWordTokenizer, TweetTokenizer, MWETokenizer
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
```

```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
text = "Natural Language Processing is really amazing! I am learning some concepts like tokenization, stemming, and lemmatization."
print("Original Text:\n", text)
```

```python
ws_tokenizer = WhitespaceTokenizer().tokenize(text)
print("Whitespace Tokens:\n", ws_tokenizer)
```

```python
punct_tokens = word_tokenize(text)
print("Punctuation based Tokens:\n", punct_tokens)
```

```python
treebank_tokenizer = TreebankWordTokenizer().tokenize(text)
print("Treebank Tokens:\n", treebank_tokenizer)
```

```python
tweet_tokenizer = TweetTokenizer().tokenize(text)
print("Tweet Tokens:\n", tweet_tokenizer)
```

```python
mwe_tokenizer = MWETokenizer([('New', 'York'), ('machine', 'learning')])

sample_text = "I love machine learning and visiting New York."

mwe_tokens = mwe_tokenizer.tokenize(sample_text.split())

print("MWE Tokens:\n", mwe_tokens)
```

```python
porter = PorterStemmer()

words = word_tokenize(text)

porter_stems = [porter.stem(word) for word in words]

print("Porter Stemmer Output:\n", porter_stems)
```

```python
snowball = SnowballStemmer("english")

snowball_stems = [snowball.stem(word) for word in words]

print("Snowball Stemmer Output:\n", snowball_stems)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in words]

print("Lemmatized Words:\n", lemmatized_words)
```
