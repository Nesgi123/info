```python
import pandas as pd
```

```python
# -------------------------------
# Function for Morphological Rule
# -------------------------------
def add_delete_rule(base, add="", delete_suffix=""):
    """
    base = original word
    add = string to add
    delete_suffix = string to remove from end
    """
    if delete_suffix and base.endswith(delete_suffix):
        new_word = base[:-len(delete_suffix)] + add
    else:
        new_word = base + add
    return new_word
```

```python
# -------------------------------
# Examples
# -------------------------------
examples = [
    ("play", "ing", ""),
    ("walk", "ed", ""),
    ("cat", "s", ""),
    ("try", "ies", "y"),
    ("happy", "ness", ""),
    ("legal", "illegal", "")   # special case
]
```

```python
# -------------------------------
# Create Add-Delete Table
# -------------------------------
data = []

for base, add, delete in examples:
    result = add_delete_rule(base, add, delete)
    data.append([base, add, delete if delete else "-", result])

df = pd.DataFrame(data, columns=["Base Word", "Add", "Delete", "Result Word"])

print(df)
```

```python
def morphology(base, prefix="", suffix="", delete_suffix=""):
    if delete_suffix and base.endswith(delete_suffix):
        base = base[:-len(delete_suffix)]
    return prefix + base + suffix
```

```python
examples2 = [
    ("happy", "un", "", ""),
    ("possible", "im", "", ""),
    ("regular", "ir", "", ""),
    ("use", "", "ful", ""),
    ("act", "", "ion", "")
]
```

```python
data2 = []

for base, prefix, suffix, delete in examples2:
    result = morphology(base, prefix, suffix, delete)
    data2.append([
        base,
        prefix if prefix else "-",
        suffix if suffix else "-",
        delete if delete else "-",
        result
    ])

df2 = pd.DataFrame(data2,
                   columns=["Base", "Prefix Add", "Suffix Add", "Delete", "Result"])

print(df2)
```
